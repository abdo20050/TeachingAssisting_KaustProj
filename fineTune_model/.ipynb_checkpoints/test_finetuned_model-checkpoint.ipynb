```python
torch.cuda.empty_cache()
```

```python
encoded_input = tokenizer(text, padding = True,truncation = True, max_length = 1024,return_tensors="pt").to(device)
```

```python
model.eval()
encoded_output = model.generate(**encoded_input,min_length = 64, max_length = 256)
```

```python
output_txt = tokenizer.batch_decode(encoded_output, skip_special_tokens=True)[0]
```
